```python
from qiskit import QuantumCircuit

from qiskit.quantum_info import Statevector



def build_cnot_demo(control_value):

    circuit = QuantumCircuit(2)

    if control_value == 1:

        circuit.x(0)

    circuit.cx(0, 1)

    return circuit



for control_val in [1, 0]:

    circuit = build_cnot_demo(control_val)

    sv = Statevector.from_instruction(circuit)

    print(f"Control qubit initialised to |{control_val}>")

    print(sv)

    print()
```

```python
from qiskit import QuantumCircuit, transpile

from qiskit_aer import AerSimulator



def make_bell_circuit():

    qc = QuantumCircuit(2, 2)

    qc.h(0)

    qc.cx(0, 1)

    qc.measure([0, 1], [0, 1])

    return qc



backend = AerSimulator()

circuit = make_bell_circuit()

compiled = transpile(circuit, backend)

job = backend.run(compiled, shots=1024)

counts = job.result().get_counts()

print("Bell state measurement counts:", counts)
```

```python
from qiskit import QuantumCircuit, transpile

from qiskit_aer import AerSimulator



def make_ghz_circuit(n_qubits):

    qc = QuantumCircuit(n_qubits, n_qubits)

    qc.h(0)

    for i in range(n_qubits - 1):

        qc.cx(i, i + 1)

    qc.measure(range(n_qubits), range(n_qubits))

    return qc



backend = AerSimulator()

circuit = make_ghz_circuit(3)

compiled = transpile(circuit, backend)

job = backend.run(compiled, shots=1024)

counts = job.result().get_counts()

print("3-qubit GHZ measurement counts:", counts)
```

```python
from qiskit import QuantumCircuit

from qiskit.quantum_info import Statevector



def build_full_adder(a_bit, b_bit):

    qc = QuantumCircuit(3)

    if a_bit:

        qc.x(0)

    if b_bit:

        qc.x(1)

    qc.cx(0, 2)   # Sum: XOR contribution from A

    qc.cx(1, 2)   # Sum: XOR contribution from B

    qc.ccx(0, 1, 2)  # Carry-style AND term

    return qc



a_bit, b_bit = 1, 0

circuit = build_full_adder(a_bit, b_bit)

sv = Statevector.from_instruction(circuit)

print(f"Inputs: A={a_bit}, B={b_bit}, Carry-in=0")

print(sv)
```

```python
from qiskit import QuantumCircuit

from qiskit.quantum_info import Statevector

import numpy as np



def build_custom_entangled_state(angle):

    qc = QuantumCircuit(2)

    qc.ry(angle, 0)

    qc.cx(0, 1)

    qc.s(1)

    return qc



angle = np.pi / 5

circuit = build_custom_entangled_state(angle)

sv = Statevector.from_instruction(circuit)

print("Statevector:")

print(sv)



print("\nProbability distribution:")

probs = np.abs(sv.data) ** 2

for idx, p in enumerate(probs):

    bitstring = format(idx, "02b")

    print(f"  |{bitstring}> : {p:.4f}")
```
